```python
encoding = tiktoken.encoding_for_model("gpt-5")
```

```python
for i in range(0,30) :
    token_text = encoding.decode([i])
    print(token_text)
```

```python
tokens = encoding.encode("My name is Shar")
print(tokens)
for token_id in tokens :
     
    token_text = encoding.decode([token_id])

    print(token_text)
```
