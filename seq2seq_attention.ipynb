```python
# Create a beginner-friendly short lenght subset
# English sentence ≤ 10 words French sentence  ≤ 15 words

short_pairs = []

for example in dataset["train"]:

    english = example["translation"]["en"].strip()
    french = example["translation"]["fr"].strip()

    english_length = len(english.split())
    french_length = len(french.split())

    if english_length <= 10 and french_length <= 15:
        short_pairs.append((english, french))


print("Total original pairs:", len(dataset["train"]))
print("Short sentence pairs:", len(short_pairs))
```

```python
# Clean and tokenize the sentences
import re


def normalize_text(text):

    # Convert everything to lowercase
    text = text.lower()

    # Convert curly apostrophes to normal apostrophe
    text = text.replace("’", "'")
    text = text.replace("‘", "'")

    # Normalize some common dash characters
    text = text.replace("–", "-")
    text = text.replace("—", "-")

    # Remove leading/trailing whitespace
    text = text.strip()

    return text
```

```python
text = "What's the Matter?’"

cleaned = normalize_text(text)

print("Original:", text)
print("Cleaned :", cleaned)
```

```python
def tokenize(text):

    text = normalize_text(text)

    tokens = re.findall(
        r"\w+|[^\w\s]",  # \w+ → matches words (letters, digits, underscore).[^\w\s] → matches punctuation and symbols (anything that’s not a word character or whitespace).
        text,
        flags=re.UNICODE
    )

    return tokens
```

So far, we have created two completely separate vocabularies: English vocabulary ,French vocabulary

Each vocabulary is basically: word → integer ID

For example, your English vocabulary might contain:
<PAD>     → 0
<UNK>     → 1
<START>   → 2
<END>     → 3
the       → 4
i         → 5
house     → 25
happy     → 47

while the French vocabulary might independently contain:
<PAD>     → 0
<UNK>     → 1
<START>   → 2
<END>     → 3
de        → 4
je        → 5
maison    → 81
heureux   → 136

Note : There is no mapping or relation between Id of English and French . English ID 25 has absolutely no requirement to correspond to French ID 25.



```python
import torch


def prepare_training_pair(english_tokens, french_tokens):

    # -----------------------------
    # 1. Prepare encoder input
    # -----------------------------

    # Add <END> to the English sentence
    encoder_tokens = english_tokens + ["<END>"]

    # Convert English words into English vocabulary IDs
    encoder_ids = [
        english_word_to_id.get(
            word,
            english_word_to_id["<UNK>"]
        )
        for word in encoder_tokens
    ]


    # -----------------------------
    # 2. Prepare decoder input
    # -----------------------------

    # Decoder starts with <START>
    decoder_input_tokens = (
        ["<START>"] + french_tokens
    )

    # Convert French words into French vocabulary IDs
    decoder_input_ids = [
        french_word_to_id.get(
            word,
            french_word_to_id["<UNK>"]
        )
        for word in decoder_input_tokens
    ]


    # -----------------------------
    # 3. Prepare decoder target
    # -----------------------------

    # Decoder should finally predict <END>
    decoder_target_tokens = (
        french_tokens + ["<END>"]
    )

    # Convert target words into French vocabulary IDs
    decoder_target_ids = [
        french_word_to_id.get(
            word,
            french_word_to_id["<UNK>"]
        )
        for word in decoder_target_tokens
    ]


    # -----------------------------
    # 4. Convert Python lists
    #    into PyTorch tensors
    # -----------------------------

    encoder_tensor = torch.tensor(
        encoder_ids,
        dtype=torch.long
    )

    decoder_input_tensor = torch.tensor(
        decoder_input_ids,
        dtype=torch.long
    )

    decoder_target_tensor = torch.tensor(
        decoder_target_ids,
        dtype=torch.long
    )


    return (
        encoder_tensor,
        decoder_input_tensor,
        decoder_target_tensor
    )
```

##Build the LSTM Encoder

This will be similar to the encoder from your date project, with one very important difference.

In the date project, we mainly cared about:

final hidden state h
final cell state c

Now, because we are going to introduce attention, we also need to preserve:

h1, h2, h3, ..., hN

the encoder output after every English token.

```python
# Build the LSTM Encoder

import torch
import torch.nn as nn

class Encoder(nn.Module):

    def __init__(self, vocab_size, embedding_size, hidden_size):

        super().__init__()

        # Convert each English word ID into a learned vector
        self.embedding = nn.Embedding(
            num_embeddings=vocab_size,
            embedding_dim=embedding_size
        )

        # LSTM processes the English word embeddings one by one
        self.lstm = nn.LSTM(
            input_size=embedding_size,
            hidden_size=hidden_size
        )


    def forward(self, encoder_input):

        # Example:
        # [885, 24, 479, ...]
        #       ↓
        # each ID becomes an embedding vector
        embedded = self.embedding(encoder_input)


        # Pass the whole sequence through the LSTM
        encoder_outputs, (hidden_state, cell_state) = self.lstm(
            embedded
        )


        # encoder_outputs contains h1, h2, h3 ... hN
        # hidden_state contains final hN
        # cell_state contains final cN

        return encoder_outputs, hidden_state, cell_state

```

embedding_size = 64.Each English word gets represented by 64 learned numbers.
Those 64 numbers start randomly and are learned during training.

hidden_size = 128

At every time step, the LSTM produces a hidden representation containing:

128 numbers

So:

during → h1 = [128 numbers]

this   → h2 = [128 numbers]

time   → h3 = [128 numbers]

...

<END>  → h12 = [128 numbers]

Attention part

We will use Bahdanau / additive attention, one of the classic attention mechanisms used with RNN/LSTM encoder-decoders.

```python
class Attention(nn.Module):

    def __init__(self, hidden_size):

        super().__init__()

        # Transform encoder hidden states
        self.encoder_layer = nn.Linear(
            hidden_size,
            hidden_size
        )

        # Transform decoder hidden state
        self.decoder_layer = nn.Linear(
            hidden_size,
            hidden_size
        )

        # Convert the combined information
        # into one attention score
        self.score_layer = nn.Linear(
            hidden_size,
            1
        )


    def forward(self, decoder_hidden, encoder_outputs):

        # decoder_hidden shape:
        # [1, hidden_size]

        # encoder_outputs shape:
        # [sequence_length, hidden_size]


        # Remove the first dimension
        # [1, 128] -> [128]
        decoder_hidden = decoder_hidden.squeeze(0)


        # Transform all encoder states
        encoder_part = self.encoder_layer(
            encoder_outputs
        )


        # Transform decoder state
        decoder_part = self.decoder_layer(
            decoder_hidden
        )


        # Compare each encoder state
        # with the current decoder state
        combined = torch.tanh(
            encoder_part + decoder_part
        )


        # Produce one score for every
        # encoder time step
        scores = self.score_layer(
            combined
        ).squeeze(1)


        # Convert scores into probabilities
        # that add up to 1
        attention_weights = torch.softmax(
            scores,
            dim=0
        )


        # Multiply each encoder state by
        # its attention weight
        weighted_encoder_outputs = (
            attention_weights.unsqueeze(1)
            *
            encoder_outputs
        )


        # Add them together to create
        # one context vector
        context_vector = torch.sum(
            weighted_encoder_outputs,
            dim=0,
            keepdim=True
        )


        return context_vector, attention_weights
```

Decoder section 

```python
class AttentionDecoder(nn.Module):

    def __init__(
        self,
        vocab_size,
        embedding_size,
        hidden_size
    ):

        super().__init__()

        # Convert each French word ID into a learned vector
        self.embedding = nn.Embedding(
            num_embeddings=vocab_size,
            embedding_dim=embedding_size
        )

        # Attention mechanism
        self.attention = Attention(
            hidden_size=hidden_size
        )

        # Decoder receives:
        # French embedding + attention context
        self.lstm = nn.LSTM(
            input_size=embedding_size + hidden_size,
            hidden_size=hidden_size
        )

        # Convert decoder hidden state
        # into scores for every French word
        self.output_layer = nn.Linear(
            hidden_size,
            vocab_size
        )


    def forward(
        self,
        current_token,
        hidden_state,
        cell_state,
        encoder_outputs
    ):

        # ---------------------------------
        # 1. Embed current French token
        # ---------------------------------

        embedded = self.embedding(current_token)


        # ---------------------------------
        # 2. Calculate attention
        # ---------------------------------

        context_vector, attention_weights = (
            self.attention(
                hidden_state,
                encoder_outputs
            )
        )


        # ---------------------------------
        # 3. Combine French embedding
        #    with attention context
        # ---------------------------------

        decoder_input = torch.cat(
            (embedded, context_vector),
            dim=1
        )


        # ---------------------------------
        # 4. Pass combined information
        #    through decoder LSTM
        # ---------------------------------

        decoder_output, (
            hidden_state,
            cell_state
        ) = self.lstm(
            decoder_input,
            (hidden_state, cell_state)
        )


        # ---------------------------------
        # 5. Produce scores for all
        #    possible French words
        # ---------------------------------

        logits = self.output_layer(
            decoder_output
        )


        return (
            logits,
            hidden_state,
            cell_state,
            attention_weights
        )
```

```python
# Decoder starts with the encoder's final memory
decoder_hidden = encoder_hidden
decoder_cell = encoder_cell


# We will store the output scores from every decoder step
all_logits = []

# We will also store attention weights
# so we can inspect them later
all_attention_weights = []


# Go through each decoder input token one by one
for step in range(len(decoder_input_tensor)):

    # Get one French input token
    #
    # Example:
    # step 0 -> <START>
    # step 1 -> pendant
    # step 2 -> ce
    current_token = decoder_input_tensor[step].unsqueeze(0)


    # Run one attention-decoder step
    (
        logits,
        decoder_hidden,
        decoder_cell,
        attention_weights
    ) = decoder(
        current_token,
        decoder_hidden,
        decoder_cell,
        encoder_outputs
    )


    # Store the prediction scores
    all_logits.append(
        logits.squeeze(0)
    )


    # Store the attention weights
    all_attention_weights.append(
        attention_weights
    )
```

```python
# Combine predictions from all decoder steps
all_logits = torch.stack(all_logits)

print("All logits shape:")
print(all_logits.shape)

print("\nTarget tensor shape:")
print(decoder_target_tensor.shape)
```

```python
# Find the highest-scoring vocabulary ID
# at every decoder step
predicted_ids = torch.argmax(
    all_logits,
    dim=1
)


# Convert predicted IDs back into French words
predicted_words = [
    french_id_to_word[word_id.item()]
    for word_id in predicted_ids
]


print("Predicted words:")
print(predicted_words)

print("\nCorrect target words:")
print(
    french_tokens + ["<END>"]
)
```
